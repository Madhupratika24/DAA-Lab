```python
def is_safe(board, row, col):
    # Check previous rows for conflicts
    for prev_row in range(row):
        placed = board[prev_row]

        if placed == col:  # Same column
            return False

        if abs(prev_row - row) == abs(placed - col):  # Same diagonal
            return False

    return True


def solve_n_queens(n):
    board = [-1] * n
    solutions = []
    backtrack_count = [0]

    def backtrack(row):
        if row == n:
            solutions.append(board[:])
            return

        for col in range(n):
            if is_safe(board, row, col):
                board[row] = col

                backtrack(row + 1)

                board[row] = -1  # Undo choice
                backtrack_count[0] += 1

    backtrack(0)

    return solutions, backtrack_count[0]


def display_board(solution, n):
    print(' +' + '---+' * n)

    for row in range(n):
        print(' |', end='')

        for col in range(n):
            if solution[row] == col:
                print(' Q |', end='')
            else:
                print(' . |', end='')

        print()

    print(' +' + '---+' * n)


# Solve for N = 4, 6, and 8
for n in [4, 6, 8]:

    solutions, backtracks = solve_n_queens(n)

    print(f"\nN={n}: {len(solutions)} solutions, {backtracks} backtracks")

    # Display all solutions only for N=4
    if n == 4:
        print(f"\nAll solutions for {n}-Queens:")

        for i, sol in enumerate(solutions, 1):
            print(f"\nSolution {i}: {sol}")
            display_board(sol, n)
```

```python
# --------------------------------------------------
# Traditional N-Queens Backtracking (Loop Checking)
# --------------------------------------------------

def solve_n_queens_normal(n):
    board = [-1] * n
    solutions = 0
    backtracks = 0

    def is_safe(row, col):
        for prev_row in range(row):
            placed_col = board[prev_row]

            if placed_col == col:
                return False

            if abs(prev_row - row) == abs(placed_col - col):
                return False

        return True

    def backtrack(row):
        nonlocal solutions, backtracks

        if row == n:
            solutions += 1
            return

        for col in range(n):
            if is_safe(row, col):
                board[row] = col
                backtrack(row + 1)
                board[row] = -1
                backtracks += 1

    backtrack(0)

    return solutions, backtracks


# --------------------------------------------------
# Bitmask Optimized N-Queens Backtracking
# --------------------------------------------------

def solve_n_queens_bitmask(n):

    solutions = 0
    backtracks = 0

    # n bits set to 1
    full_mask = (1 << n) - 1

    def backtrack(row, columns, diag1, diag2):
        nonlocal solutions, backtracks

        if row == n:
            solutions += 1
            return

        # Available positions
        available = full_mask & ~(columns | diag1 | diag2)

        while available:

            # Get rightmost available position
            position = available & -available

            # Remove this position
            available -= position

            backtrack(
                row + 1,
                columns | position,
                (diag1 | position) << 1,
                (diag2 | position) >> 1
            )

            backtracks += 1

    backtrack(0, 0, 0, 0)

    return solutions, backtracks


# --------------------------------------------------
# Compare for N = 12
# --------------------------------------------------

n = 12

normal_solutions, normal_backtracks = solve_n_queens_normal(n)

bit_solutions, bit_backtracks = solve_n_queens_bitmask(n)


print("N =", n)

print("\nNormal Backtracking:")
print("Solutions :", normal_solutions)
print("Backtracks:", normal_backtracks)


print("\nBitmask Backtracking:")
print("Solutions :", bit_solutions)
print("Backtracks:", bit_backtracks)


if normal_solutions == bit_solutions:
    print("\nVerification: Both methods produce the same number of solutions.")
else:
    print("\nVerification Failed!")
```

```python
# Knight's Tour using Backtracking

N = 8

# Possible knight moves
moves_x = [2, 1, -1, -2, -2, -1, 1, 2]
moves_y = [1, 2, 2, 1, -1, -2, -2, -1]


def is_safe(x, y, board):
    # Check if the position is inside the board and unvisited
    return (
        0 <= x < N and
        0 <= y < N and
        board[x][y] == -1
    )


def solve_knight_tour(x, y, move_number, board):
    # All cells are visited
    if move_number == N * N:
        return True

    # Try all possible knight moves
    for i in range(8):
        next_x = x + moves_x[i]
        next_y = y + moves_y[i]

        if is_safe(next_x, next_y, board):

            # Mark the move
            board[next_x][next_y] = move_number

            # Recursive call
            if solve_knight_tour(
                next_x,
                next_y,
                move_number + 1,
                board
            ):
                return True

            # Backtrack
            board[next_x][next_y] = -1

    return False


def print_solution(board):
    print("\nKnight's Tour Solution:\n")

    for row in board:
        for cell in row:
            print(f"{cell:2}", end=" ")
        print()


# ----------------------------
# Main Program
# ----------------------------

# Create empty chessboard
board = [[-1 for _ in range(N)] for _ in range(N)]

# Starting position (0,0)
start_x = 0
start_y = 0

# Mark first position
board[start_x][start_y] = 0


if solve_knight_tour(start_x, start_y, 1, board):
    print_solution(board)
else:
    print("No solution exists")
```
