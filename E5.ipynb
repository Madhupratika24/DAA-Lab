```python
import random
comparison_count = 0  # Global counter
def min_max_dc(arr, low, high):
    global comparison_count
    # Base case: single element
    if low == high:
        return arr[low], arr[low]
    # Base case: two elements
    if high == low + 1:
        comparison_count += 1
        if arr[low] < arr[high]:
            return arr[low], arr[high]
        return arr[high], arr[low]
    # Divide
    mid = (low + high) // 2
    lmin, lmax = min_max_dc(arr, low, mid)
    rmin, rmax = min_max_dc(arr, mid + 1, high)
    # Conquer: combine with 2 comparisons
    comparison_count += 1
    overall_min = lmin if lmin < rmin else rmin
    comparison_count += 1
    overall_max = lmax if lmax > rmax else rmax
    return overall_min, overall_max
def min_max_naive(arr):
    mn, mx = arr[0], arr[0]
    comps = 0
    for x in arr[1:]:
        comps += 1
        if x < mn: mn = x
        comps += 1
        if x > mx: mx = x
    return mn, mx, comps
# --- Demonstration on small array --
arr = [3, 1, 7, 4, 9, 2, 8, 5, 6, 0]
comparison_count = 0
mn, mx = min_max_dc(arr, 0, len(arr) - 1)
dc_comps = comparison_count
_, _, naive_comps = min_max_naive(arr)
print(f'Array: {arr}')
print(f'Min: {mn}, Max: {mx}')
print(f'D&C Comparisons: {dc_comps}')
print(f'Naive Comparisons: {naive_comps}')
# --- Performance Analysis --
print(f'\n{"Size":>8} {"DC Comps":>12} {"Naive Comps":>14} {"Formula 3n/2-2":>16}')
print('-' * 56)
for size in [10, 100, 1000, 10000]:
    arr = [random.randint(1, 10000) for _ in range(size)]
    comparison_count = 0
    mn, mx = min_max_dc(arr, 0, len(arr) - 1)
    dc = comparison_count
    _, _, naive = min_max_naive(arr)

    formula = 3 * size // 2 - 2
    print(f'{size:>8} {dc:>12} {naive:>14} {formula:>16}')
```

```python
import random

comparison_count = 0  # Global counter

def second_min_max_dc(arr, low, high):
    global comparison_count

    # Base case: single element
    if low == high:
        return arr[low], None, arr[low], None

    # Base case: two elements
    if high == low + 1:
        comparison_count += 1
        if arr[low] < arr[high]:
            return arr[low], arr[high], arr[high], arr[low]
        else:
            return arr[high], arr[low], arr[low], arr[high]

    # Divide
    mid = (low + high) // 2
    lmin, lsecond_min, lmax, lsecond_max = second_min_max_dc(arr, low, mid)
    rmin, rsecond_min, rmax, rsecond_max = second_min_max_dc(arr, mid + 1, high)

    # Conquer: Find Minimum and Second Minimum
    comparison_count += 1
    if lmin < rmin:
        overall_min = lmin
        candidates = [x for x in [lsecond_min, rmin] if x is not None]
        comparison_count += 1
        overall_second_min = min(candidates)
    else:
        overall_min = rmin
        candidates = [x for x in [rsecond_min, lmin] if x is not None]
        comparison_count += 1
        overall_second_min = min(candidates)

    # Conquer: Find Maximum and Second Maximum
    comparison_count += 1
    if lmax > rmax:
        overall_max = lmax
        candidates = [x for x in [lsecond_max, rmax] if x is not None]
        comparison_count += 1
        overall_second_max = max(candidates)
    else:
        overall_max = rmax
        candidates = [x for x in [rsecond_max, lmax] if x is not None]
        comparison_count += 1
        overall_second_max = max(candidates)

    return overall_min, overall_second_min, overall_max, overall_second_max


def min_max_dc(arr, low, high):
    global comparison_count

    if low == high:
        return arr[low], arr[low]

    if high == low + 1:
        comparison_count += 1
        if arr[low] < arr[high]:
            return arr[low], arr[high]
        return arr[high], arr[low]

    mid = (low + high) // 2
    lmin, lmax = min_max_dc(arr, low, mid)
    rmin, rmax = min_max_dc(arr, mid + 1, high)

    comparison_count += 1
    overall_min = lmin if lmin < rmin else rmin

    comparison_count += 1
    overall_max = lmax if lmax > rmax else rmax

    return overall_min, overall_max


# --- Demonstration on small array ---
arr = [3, 1, 7, 4, 9, 2, 8, 5, 6, 0]

comparison_count = 0
mn, second_mn, mx, second_mx = second_min_max_dc(arr, 0, len(arr) - 1)
extended_comps = comparison_count

comparison_count = 0
_, _ = min_max_dc(arr, 0, len(arr) - 1)
normal_comps = comparison_count

print(f'Array: {arr}')
print(f'Min: {mn}')
print(f'Second Min: {second_mn}')
print(f'Max: {mx}')
print(f'Second Max: {second_mx}')
print(f'Extended Comparisons: {extended_comps}')
print(f'Min-Max Comparisons: {normal_comps}')
print(f'Additional Comparisons: {extended_comps - normal_comps}')

# --- Performance Analysis ---
print(f'\n{"Size":>8} {"Min-Max":>12} {"Extended":>12} {"Additional":>12}')
print('-' * 50)

for size in [10, 100, 1000, 10000]:
    arr = [random.randint(1, 10000) for _ in range(size)]

    comparison_count = 0
    second_min_max_dc(arr, 0, len(arr) - 1)
    extended = comparison_count

    comparison_count = 0
    min_max_dc(arr, 0, len(arr) - 1)
    normal = comparison_count

    print(f'{size:>8} {normal:>12} {extended:>12} {extended-normal:>12}')
```

```python
import random

comparison_count = 0  # Global counter

def partition(arr, low, high):
    global comparison_count

    # Choose last element as pivot
    pivot = arr[high]
    i = low - 1

    for j in range(low, high):
        comparison_count += 1
        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]

    arr[i + 1], arr[high] = arr[high], arr[i + 1]

    return i + 1


def quick_select(arr, low, high, k):
    # Base case: single element
    if low == high:
        return arr[low]

    # Divide
    pivot_index = partition(arr, low, high)

    # Position of pivot
    position = pivot_index - low + 1

    # Conquer
    if k == position:
        return arr[pivot_index]

    elif k < position:
        return quick_select(arr, low, pivot_index - 1, k)

    else:
        return quick_select(arr, pivot_index + 1, high, k - position)


# --- Demonstration on small array ---
arr = [45, 12, 78, 3, 56, 23, 89, 10, 67, 34]

print(f'Array: {arr}')

for k in [1, len(arr), len(arr)//2]:
    comparison_count = 0
    result = quick_select(arr.copy(), 0, len(arr)-1, k)
    print(f'{k}-th Smallest Element: {result}')
    print(f'Comparisons: {comparison_count}')


# --- Performance Analysis on 1000 Random Integers ---
print("\n--- Performance Analysis ---")

arr = [random.randint(1, 10000) for _ in range(1000)]

print(f'{"K Value":>10} {"Element":>12} {"Comparisons":>15}')
print('-' * 40)

for k in [1, len(arr), len(arr)//2]:
    comparison_count = 0

    result = quick_select(arr.copy(), 0, len(arr)-1, k)

    print(f'{k:>10} {result:>12} {comparison_count:>15}')
```
