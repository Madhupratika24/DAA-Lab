```python
import heapq
from itertools import permutations

INF = float('inf')


def reduce_matrix(mat):
    """Reduce matrix and return reduction cost"""

    import copy

    m = [row[:] for row in mat]
    n = len(m)
    cost = 0

    # Row reduction
    for i in range(n):
        row_min = min(m[i])

        if row_min and row_min != INF:
            cost += row_min
            m[i] = [
                x - row_min if x != INF else INF
                for x in m[i]
            ]

    # Column reduction
    for j in range(n):
        col_min = min(m[i][j] for i in range(n))

        if col_min and col_min != INF:
            cost += col_min

            for i in range(n):
                if m[i][j] != INF:
                    m[i][j] -= col_min

    return m, cost


def tsp_brute_force(cost, n):
    """Brute force solution for verification"""

    cities = list(range(1, n))

    best_cost = INF
    best_path = None

    for perm in permutations(cities):

        path = [0] + list(perm) + [0]

        c = sum(
            cost[path[i]][path[i + 1]]
            for i in range(n)
        )

        if c < best_cost:
            best_cost = c
            best_path = path

    return best_path, best_cost


# --- 5-city cost matrix ---

cost = [
    [INF, 10, 8, 9, 7],
    [10, INF, 10, 5, 6],
    [8, 10, INF, 8, 9],
    [9, 5, 8, INF, 6],
    [7, 6, 9, 6, INF]
]

n = 5

cities = ['A', 'B', 'C', 'D', 'E']


# Find optimal tour
best_path, best_cost = tsp_brute_force(cost, n)


# Display cost matrix
print("5-City TSP - Cost Matrix:")

print(
    f'{"":>4}',
    ' '.join(f'{c:>5}' for c in cities)
)

for i, row in enumerate(cost):

    r = [
        'INF' if x == INF else str(x)
        for x in row
    ]

    print(
        f'{cities[i]:>4}',
        ' '.join(f'{v:>5}' for v in r)
    )


# Display result
print(
    f'\nOptimal Tour: '
    f'{" -> ".join(cities[i] for i in best_path)}'
)

print(f'Minimum Cost: {best_cost}')


# Path verification
print("\nPath verification:")

for i in range(n):
    u = best_path[i]
    v = best_path[i + 1]

    print(
        f' {cities[u]} -> {cities[v]}: '
        f'cost = {cost[u][v]}'
    )
```

```python
from itertools import permutations

INF = float('inf')


# ---------------------------------------
# Brute Force TSP (Optimal Solution)
# ---------------------------------------

def tsp_brute_force(distance, n):

    cities = list(range(1, n))

    best_cost = INF
    best_route = None

    for perm in permutations(cities):

        route = [0] + list(perm) + [0]

        cost = sum(
            distance[route[i]][route[i + 1]]
            for i in range(n)
        )

        if cost < best_cost:
            best_cost = cost
            best_route = route

    return best_route, best_cost



# ---------------------------------------
# Nearest Neighbour Heuristic
# ---------------------------------------

def nearest_neighbour(distance, n):

    visited = [False] * n

    route = [0]
    visited[0] = True

    total_cost = 0

    current = 0

    for _ in range(n - 1):

        nearest = INF
        next_city = -1

        for city in range(n):

            if not visited[city] and distance[current][city] < nearest:
                nearest = distance[current][city]
                next_city = city

        route.append(next_city)
        visited[next_city] = True

        total_cost += nearest
        current = next_city


    # Return to starting city
    total_cost += distance[current][0]
    route.append(0)

    return route, total_cost



# ---------------------------------------
# 6-City Road Distance Matrix (km)
# ---------------------------------------

# Cities:
# A, B, C, D, E, F

distance = [
    [INF, 10, 15, 20, 25, 18],
    [10, INF, 35, 25, 17, 30],
    [15, 35, INF, 30, 28, 20],
    [20, 25, 30, INF, 15, 22],
    [25, 17, 28, 15, INF, 12],
    [18, 30, 20, 22, 12, INF]
]


cities = ['A', 'B', 'C', 'D', 'E', 'F']

n = len(distance)



# ---------------------------------------
# Solve TSP
# ---------------------------------------

optimal_route, optimal_cost = tsp_brute_force(distance, n)

nn_route, nn_cost = nearest_neighbour(distance, n)



# ---------------------------------------
# Display Results
# ---------------------------------------

print("6-City TSP Road Distance Matrix (km):\n")

print(f'{"":>5}', end="")
for c in cities:
    print(f'{c:>6}', end="")
print()


for i in range(n):

    print(f'{cities[i]:>5}', end="")

    for j in range(n):

        if distance[i][j] == INF:
            print(f'{"INF":>6}', end="")
        else:
            print(f'{distance[i][j]:>6}', end="")

    print()



print("\n--- Brute Force Optimal Solution ---")

print(
    "Route:",
    " -> ".join(cities[i] for i in optimal_route)
)

print("Minimum Distance:", optimal_cost, "km")



print("\n--- Nearest Neighbour Solution ---")

print(
    "Route:",
    " -> ".join(cities[i] for i in nn_route)
)

print("Distance:", nn_cost, "km")



print("\n--- Comparison ---")

difference = nn_cost - optimal_cost

print(
    "Extra distance using Nearest Neighbour:",
    difference,
    "km"
)

if optimal_cost == nn_cost:
    print("Nearest Neighbour found the optimal route.")
else:
    print("Nearest Neighbour produced a non-optimal route.")
```

```python
import random
from itertools import permutations
import math

INF = float('inf')


# ------------------------------------------
# Calculate total tour distance
# ------------------------------------------

def tour_cost(route, dist):
    cost = 0

    for i in range(len(route) - 1):
        cost += dist[route[i]][route[i + 1]]

    return cost



# ------------------------------------------
# Brute Force TSP (Optimal Solution)
# ------------------------------------------

def tsp_brute_force(dist, n):

    cities = list(range(1, n))

    best_cost = INF
    best_route = None

    for perm in permutations(cities):

        route = [0] + list(perm) + [0]

        cost = tour_cost(route, dist)

        if cost < best_cost:
            best_cost = cost
            best_route = route

    return best_route, best_cost



# ------------------------------------------
# Nearest Neighbour Heuristic
# ------------------------------------------

def nearest_neighbour(dist, n):

    visited = [False] * n

    route = [0]
    visited[0] = True

    current = 0

    while len(route) < n:

        nearest = INF
        next_city = -1

        for city in range(n):

            if not visited[city] and dist[current][city] < nearest:
                nearest = dist[current][city]
                next_city = city

        route.append(next_city)
        visited[next_city] = True
        current = next_city

    route.append(0)

    return route



# ------------------------------------------
# 2-opt Improvement
# ------------------------------------------

def two_opt(route, dist):

    improved = True

    best_route = route[:]

    while improved:

        improved = False

        best_distance = tour_cost(best_route, dist)

        for i in range(1, len(route) - 2):

            for j in range(i + 1, len(route) - 1):

                # Reverse segment
                new_route = (
                    best_route[:i]
                    + best_route[i:j + 1][::-1]
                    + best_route[j + 1:]
                )

                new_distance = tour_cost(new_route, dist)

                if new_distance < best_distance:

                    best_route = new_route
                    best_distance = new_distance
                    improved = True

        route = best_route[:]

    return best_route



# ------------------------------------------
# Generate Random 10-City Problem
# ------------------------------------------

random.seed(10)

n = 10

coordinates = [
    (random.randint(0, 100),
     random.randint(0, 100))
    for _ in range(n)
]


# Create distance matrix

dist = [[0] * n for _ in range(n)]

for i in range(n):

    for j in range(n):

        if i != j:
            x1, y1 = coordinates[i]
            x2, y2 = coordinates[j]

            dist[i][j] = round(
                math.sqrt(
                    (x1-x2)**2 +
                    (y1-y2)**2
                ),
                2
            )

        else:
            dist[i][j] = INF



# ------------------------------------------
# Run Algorithms
# ------------------------------------------

print("10 City Coordinates:")
for i, c in enumerate(coordinates):
    print(f"City {i}: {c}")


# Optimal (10! is large but still manageable for this demo)
optimal_route, optimal_cost = tsp_brute_force(dist, n)


# Nearest neighbour
nn_route = nearest_neighbour(dist, n)
nn_cost = tour_cost(nn_route, dist)


# Apply 2-opt
two_opt_route = two_opt(nn_route, dist)
two_opt_cost = tour_cost(two_opt_route, dist)



# ------------------------------------------
# Results
# ------------------------------------------

print("\nOptimal Brute Force Solution:")
print("Route:", optimal_route)
print("Cost:", optimal_cost)


print("\nNearest Neighbour Solution:")
print("Route:", nn_route)
print("Cost:", nn_cost)


print("\n2-opt Improved Solution:")
print("Route:", two_opt_route)
print("Cost:", two_opt_cost)


print("\nComparison:")

gap = two_opt_cost - optimal_cost

percentage = (gap / optimal_cost) * 100

print(
    "Difference from optimal:",
    round(gap, 2),
    "km"
)

print(
    "2-opt solution is",
    round(percentage, 2),
    "% above optimal"
)
```
