```python
def matrix_chain_order(dims):
    """
    Matrix Chain Multiplication using DP

    dims: list of dimensions, matrix i has dims[i-1] x dims[i]
    Time: O(n^3), Space: O(n^2)
    """

    n = len(dims) - 1

    # m[i][j] = minimum multiplications for matrices i..j
    m = [[0] * (n + 1) for _ in range(n + 1)]

    # s[i][j] = index of optimal split
    s = [[0] * (n + 1) for _ in range(n + 1)]

    # l is the chain length
    for l in range(2, n + 1):
        for i in range(1, n - l + 2):
            j = i + l - 1
            m[i][j] = float('inf')

            for k in range(i, j):
                cost = (
                    m[i][k]
                    + m[k + 1][j]
                    + dims[i - 1] * dims[k] * dims[j]
                )

                if cost < m[i][j]:
                    m[i][j] = cost
                    s[i][j] = k

    return m, s


def print_optimal_parens(s, i, j):
    if i == j:
        return f"A{i}"

    k = s[i][j]

    left = print_optimal_parens(s, i, k)
    right = print_optimal_parens(s, k + 1, j)

    return f"({left} x {right})"


def print_dp_table(m, n):
    print("\nDP Cost Table m[i][j]:")

    print(f'{"":>6}', end='')
    for j in range(1, n + 1):
        print(f'A{j:>8}', end='')
    print()

    for i in range(1, n + 1):
        print(f'A{i:<5}', end='')

        for j in range(1, n + 1):
            if j < i:
                print(f'{"---":>9}', end='')
            else:
                print(f'{m[i][j]:>9}', end='')

        print()


# A1(10x30), A2(30x5), A3(5x60), A4(60x10)
dims = [10, 30, 5, 60, 10]

n = len(dims) - 1

print("Matrix Dimensions:")
for i in range(n):
    print(f" A{i+1}: {dims[i]} x {dims[i+1]}")


m, s = matrix_chain_order(dims)

print(f"\nMinimum scalar multiplications: {m[1][n]}")

print(
    f"Optimal parenthesization: "
    f"{print_optimal_parens(s, 1, n)}"
)

print_dp_table(m, n)
```

```python
def matrix_chain_order(dims):
    """
    Matrix Chain Multiplication using Dynamic Programming
    Time Complexity: O(n^3)
    Space Complexity: O(n^2)
    """

    n = len(dims) - 1

    # m[i][j] stores minimum multiplication cost
    m = [[0] * (n + 1) for _ in range(n + 1)]

    # s[i][j] stores optimal split position
    s = [[0] * (n + 1) for _ in range(n + 1)]

    # Chain length
    for length in range(2, n + 1):

        for i in range(1, n - length + 2):

            j = i + length - 1
            m[i][j] = float('inf')

            # Try all possible splits
            for k in range(i, j):

                cost = (
                    m[i][k]
                    + m[k + 1][j]
                    + dims[i - 1] * dims[k] * dims[j]
                )

                if cost < m[i][j]:
                    m[i][j] = cost
                    s[i][j] = k

    return m, s


def optimal_parenthesis(s, i, j):
    """
    Prints optimal matrix multiplication order
    """

    if i == j:
        return chr(64 + i)   # A, B, C...

    k = s[i][j]

    left = optimal_parenthesis(s, i, k)
    right = optimal_parenthesis(s, k + 1, j)

    return f"({left} x {right})"


# Matrix dimensions:
# T(4x4), R(4x4), S(4x4), V(4x10000), P(10000x1)

dims = [4, 4, 4, 4, 10000, 1]

matrix_names = ["T", "R", "S", "V", "P"]

n = len(dims) - 1

print("Matrices:")
for i in range(n):
    print(f"{matrix_names[i]} : {dims[i]} x {dims[i+1]}")


# Apply Matrix Chain Multiplication
m, s = matrix_chain_order(dims)


print("\nMinimum scalar multiplications:")
print(m[1][n])


print("\nOptimal multiplication order:")
print(optimal_parenthesis(s, 1, n))
```

```python
# Top-Down Memoized Matrix Chain Multiplication

def matrix_chain_memoized(dims):
    n = len(dims) - 1

    # Cache table
    memo = [[-1] * (n + 1) for _ in range(n + 1)]

    def solve(i, j):
        # Single matrix requires no multiplication
        if i == j:
            return 0

        # Return cached result
        if memo[i][j] != -1:
            return memo[i][j]

        memo[i][j] = float('inf')

        # Try all possible splits
        for k in range(i, j):
            cost = (
                solve(i, k)
                + solve(k + 1, j)
                + dims[i - 1] * dims[k] * dims[j]
            )

            if cost < memo[i][j]:
                memo[i][j] = cost

        return memo[i][j]

    return solve(1, n)


# Bottom-Up DP Matrix Chain Multiplication

def matrix_chain_bottom_up(dims):
    n = len(dims) - 1

    # DP table
    dp = [[0] * (n + 1) for _ in range(n + 1)]

    # Chain length
    for length in range(2, n + 1):

        for i in range(1, n - length + 2):

            j = i + length - 1
            dp[i][j] = float('inf')

            for k in range(i, j):

                cost = (
                    dp[i][k]
                    + dp[k + 1][j]
                    + dims[i - 1] * dims[k] * dims[j]
                )

                if cost < dp[i][j]:
                    dp[i][j] = cost

    return dp[1][n]


# -------------------------------
# Test with 6 matrices
# -------------------------------

# Matrices:
# A1: 30x35
# A2: 35x15
# A3: 15x5
# A4: 5x10
# A5: 10x20
# A6: 20x25

dims = [30, 35, 15, 5, 10, 20, 25]


print("Matrix Chain:")
for i in range(len(dims) - 1):
    print(f"A{i+1}: {dims[i]} x {dims[i+1]}")


# Calculate using both methods
memo_cost = matrix_chain_memoized(dims)
bottom_up_cost = matrix_chain_bottom_up(dims)


print("\nTop-Down Memoized DP Cost:")
print(memo_cost)

print("\nBottom-Up DP Cost:")
print(bottom_up_cost)


# Verify results
if memo_cost == bottom_up_cost:
    print("\nVerification: Both approaches give the same minimum cost.")
else:
    print("\nVerification Failed: Costs are different.")
```
