```python
def first_fit(items, capacity=1.0):
    bins = []          # Remaining space in each bin
    bin_contents = []  # Items in each bin

    for item in items:
        placed = False
        for i, space in enumerate(bins):
            if space >= item:
                bins[i] -= item
                bin_contents[i].append(item)
                placed = True
                break

        if not placed:
            bins.append(capacity - item)
            bin_contents.append([item])

    return bin_contents


def first_fit_decreasing(items, capacity=1.0):
    return first_fit(sorted(items, reverse=True), capacity)


def best_fit_decreasing(items, capacity=1.0):
    sorted_items = sorted(items, reverse=True)
    bins = []
    bin_contents = []

    for item in sorted_items:
        best_idx = -1
        best_space = float('inf')

        for i, space in enumerate(bins):
            if space >= item and space - item < best_space:
                best_space = space - item
                best_idx = i

        if best_idx >= 0:
            bins[best_idx] -= item
            bin_contents[best_idx].append(item)
        else:
            bins.append(capacity - item)
            bin_contents.append([item])

    return bin_contents


def display_bins(label, bins):
    print(f'\n{label}: {len(bins)} bins')

    for i, b in enumerate(bins, 1):
        used = sum(b)
        bar = '#' * int(used * 20)

        print(f'  Bin {i}: {[round(x,1) for x in b]} | Used: {used:.1f} '
              f'[{bar:<20}]')


items = [0.5, 0.7, 0.3, 0.9, 0.2, 0.6, 0.8, 0.4, 0.1, 0.5]
capacity = 1.0

lower_bound = -(-sum(items) // capacity)

print(f'Items: {items}')
print(f'Capacity: {capacity}')
print(f'Sum of items: {sum(items)}')
print(f'Lower bound on bins: {int(lower_bound)}')

ff_bins = first_fit(items)
ffd_bins = first_fit_decreasing(items)
bfd_bins = best_fit_decreasing(items)

display_bins('First Fit (FF)', ff_bins)
display_bins('First Fit Decreasing (FFD)', ffd_bins)
display_bins('Best Fit Decreasing (BFD)', bfd_bins)

print(f"\nSummary: Lower Bound={int(lower_bound)}, FF={len(ff_bins)}, "
      f"FFD={len(ffd_bins)}, BFD={len(bfd_bins)}")
```

```python
import random

# ---------------- First Fit ----------------
def first_fit(items, capacity=1.0):
    bins = []
    bin_contents = []

    for item in items:
        placed = False
        for i, space in enumerate(bins):
            if space >= item:
                bins[i] -= item
                bin_contents[i].append(item)
                placed = True
                break

        if not placed:
            bins.append(capacity - item)
            bin_contents.append([item])

    return bin_contents


# ---------------- First Fit Decreasing ----------------
def first_fit_decreasing(items, capacity=1.0):
    return first_fit(sorted(items, reverse=True), capacity)


# ---------------- Best Fit Decreasing ----------------
def best_fit_decreasing(items, capacity=1.0):
    items = sorted(items, reverse=True)
    bins = []
    bin_contents = []

    for item in items:
        best = -1
        min_space = float('inf')

        for i, space in enumerate(bins):
            if space >= item and (space - item) < min_space:
                min_space = space - item
                best = i

        if best != -1:
            bins[best] -= item
            bin_contents[best].append(item)
        else:
            bins.append(capacity - item)
            bin_contents.append([item])

    return bin_contents


# ---------------- Next Fit Decreasing ----------------
def next_fit_decreasing(items, capacity=1.0):
    items = sorted(items, reverse=True)

    bins = [capacity]
    bin_contents = [[]]
    current = 0

    for item in items:
        if bins[current] >= item:
            bins[current] -= item
            bin_contents[current].append(item)
        else:
            bins.append(capacity - item)
            bin_contents.append([item])
            current += 1

    return bin_contents


# ---------------- Simulation ----------------
TRIALS = 100
NUM_ITEMS = 50
CAPACITY = 1.0

ffd_total = 0
bfd_total = 0
nfd_total = 0

for _ in range(TRIALS):
    items = [round(random.uniform(0.1, 0.9), 2)
             for _ in range(NUM_ITEMS)]

    ffd_total += len(first_fit_decreasing(items, CAPACITY))
    bfd_total += len(best_fit_decreasing(items, CAPACITY))
    nfd_total += len(next_fit_decreasing(items, CAPACITY))

print("Average bins over", TRIALS, "trials")
print("----------------------------------")
print("FFD :", round(ffd_total / TRIALS, 2))
print("BFD :", round(bfd_total / TRIALS, 2))
print("NFD :", round(nfd_total / TRIALS, 2))
```
