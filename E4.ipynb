```python
import random
comparison_count = 0  # Global counter
def min_max_dc(arr, low, high):
    global comparison_count
    # Base case: single element
    if low == high:
        return arr[low], arr[low]
    # Base case: two elements
    if high == low + 1:
        comparison_count += 1
        if arr[low] < arr[high]:
            return arr[low], arr[high]
        return arr[high], arr[low]
    # Divide
    mid = (low + high) // 2
    lmin, lmax = min_max_dc(arr, low, mid)
    rmin, rmax = min_max_dc(arr, mid + 1, high)
    # Conquer: combine with 2 comparisons
    comparison_count += 1
    overall_min = lmin if lmin < rmin else rmin
    comparison_count += 1
    overall_max = lmax if lmax > rmax else rmax
    return overall_min, overall_max
def min_max_naive(arr):
    mn, mx = arr[0], arr[0]
    comps = 0
    for x in arr[1:]:
        comps += 1
        if x < mn: mn = x
        comps += 1
        if x > mx: mx = x
    return mn, mx, comps
# --- Demonstration on small array --
arr = [3, 1, 7, 4, 9, 2, 8, 5, 6, 0]
comparison_count = 0
mn, mx = min_max_dc(arr, 0, len(arr) - 1)
dc_comps = comparison_count
_, _, naive_comps = min_max_naive(arr)
print(f'Array: {arr}')
print(f'Min: {mn}, Max: {mx}')
print(f'D&C Comparisons: {dc_comps}')
print(f'Naive Comparisons: {naive_comps}')
# --- Performance Analysis --
print(f'\n{"Size":>8} {"DC Comps":>12} {"Naive Comps":>14} {"Formula 3n/2-2":>16}')
print('-' * 56)
for size in [10, 100, 1000, 10000]:
    arr = [random.randint(1, 10000) for _ in range(size)]
    comparison_count = 0
    mn, mx = min_max_dc(arr, 0, len(arr) - 1)
    dc = comparison_count
    _, _, naive = min_max_naive(arr)

    formula = 3 * size // 2 - 2
    print(f'{size:>8} {dc:>12} {naive:>14} {formula:>16}')
```

```python
import heapq

def dijkstra(graph, start):
    n = len(graph)
    dist = [float('inf')] * n
    parent = [-1] * n
    visited = [False] * n

    dist[start] = 0
    pq = [(0, start)]  # (distance, node)

    while pq:
        current_dist, u = heapq.heappop(pq)

        if visited[u]:
            continue
        visited[u] = True

        for v, weight in graph[u]:
            if not visited[v] and dist[u] + weight < dist[v]:
                dist[v] = dist[u] + weight
                parent[v] = u
                heapq.heappush(pq, (dist[v], v))

    return dist, parent


def get_path(parent, target):
    path = []
    while target != -1:
        path.append(target)
        target = parent[target]
    return path[::-1]


# --- 5-Node City Road Network ---
graph = [
    [(1, 10), (2, 3)],          # Airport (0)
    [(2, 1), (3, 2)],           # Node 1
    [(1, 4), (3, 8), (4, 2)],   # Node 2
    [(4, 7)],                   # Node 3
    []                          # Central Station (4)
]

start = 0      # Airport
destination = 4  # Central Station

dist, parent = dijkstra(graph, start)
path = get_path(parent, destination)

print(f"Fastest Route: {' -> '.join(map(str, path))}")
print(f"Minimum Travel Time: {dist[destination]} minutes")

# --- Distance to All Nodes ---
print("\nShortest Travel Time from Airport:")
for i in range(len(graph)):
    print(f"Node {i}: {dist[i]} minutes")
```

```python
import heapq

def dijkstra(graph, start):
    n = len(graph)

    # Distance, hop count, and parent arrays
    dist = [float('inf')] * n
    hops = [float('inf')] * n
    parent = [-1] * n

    dist[start] = 0
    hops[start] = 0

    # (distance, hop_count, vertex)
    pq = [(0, 0, start)]

    while pq:
        current_dist, current_hops, u = heapq.heappop(pq)

        # Ignore outdated entries
        if current_dist > dist[u] or (current_dist == dist[u] and current_hops > hops[u]):
            continue

        for v, weight in graph[u]:
            new_dist = dist[u] + weight
            new_hops = hops[u] + 1

            # Update if shorter path OR same distance with fewer hops
            if (new_dist < dist[v]) or (new_dist == dist[v] and new_hops < hops[v]):
                dist[v] = new_dist
                hops[v] = new_hops
                parent[v] = u
                heapq.heappush(pq, (new_dist, new_hops, v))

    return dist, hops, parent


def get_path(parent, target):
    path = []
    while target != -1:
        path.append(target)
        target = parent[target]
    return path[::-1]


# --- Graph with 6 Vertices ---
graph = [
    [(1, 4), (2, 2)],          # 0
    [(3, 5), (4, 10)],         # 1
    [(1, 2), (3, 8), (4, 4)],  # 2
    [(5, 6)],                  # 3
    [(5, 6)],                  # 4
    []                         # 5
]

start = 0
destination = 5

dist, hops, parent = dijkstra(graph, start)
path = get_path(parent, destination)

print(f"Shortest Path: {' -> '.join(map(str, path))}")
print(f"Minimum Distance: {dist[destination]}")
print(f"Hop Count: {hops[destination]}")

# --- Distance and Hop Count to All Vertices ---
print("\nVertex\tDistance\tHops")
for i in range(len(graph)):
    print(f"{i}\t{dist[i]}\t\t{hops[i]}")
```
